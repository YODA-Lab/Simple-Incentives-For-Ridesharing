```python
import pandas as pd
import plotly.express as px


results_df = pd.read_csv('Results/GIFF1000.csv')
# results_df = pd.read_csv('Results/GIFF200.csv')
```

```python
results_df.drop_duplicates(inplace=True)
results_df['alpha'] = results_df['alpha'].astype(str)
results_df['expt_type'] = results_df['giff'].astype(str)+results_df['alpha'] + '_'+results_df['alpha_d'].astype(str)
```

```python
# select delta=0, alpha!=0.1
results_df.sort_values(by='beta', inplace=True)
passenger_results_df = results_df[(results_df['delta']==0) & (results_df['alpha']!='0.1')]
# rename expt_type
names = {
    'False0.0_0.0' : 'SIP(+)',
    'True0.0_0.0' : 'GIFF',
    'True0.0_1.0' : 'GIFF(+)',
    'False1.0_0.0' : 'SIP'
}
passenger_results_df['expt_type'] = passenger_results_df['expt_type'].map(names)

fig = px.line(passenger_results_df, x='Metric (P)', y='SR', color='expt_type', hover_data=['beta'], title='GIFF1000', markers=True)
# resize
fig.update_layout(width=800, height=600)
fig.show()
```

```python
# same for drivers, except on hover, show delta 

results_df.sort_values(by='delta', inplace=True)
driver_results_df = results_df[(results_df['beta']==0) & (results_df['alpha']!='0.1')]
names = {
    'False0.0_0.0' : 'SID(+)',
    'True0.0_0.0' : 'GIFF(Q)',
    'True0.0_1.0' : 'GIFF(+)',
    'False0.0_1.0' : 'SID',
    'True0.0_0.5' : 'GIFF(R)'
}
driver_results_df['expt_type'] = driver_results_df['expt_type'].map(names)

fig = px.line(driver_results_df, x='Metric (D)', y='SR', color='expt_type', hover_data=['delta'], title='GIFF1000', markers=True)

# resize
fig.update_layout(width=800, height=600)
fig.show()
```
